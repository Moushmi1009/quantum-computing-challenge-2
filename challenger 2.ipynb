```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create simulator
simulator = AerSimulator()

# =================================
# Circuit A: 0 occurs more frequently
# =================================
circuit_A = QuantumCircuit(1, 1)
circuit_A.ry(0.5, 0)
circuit_A.measure(0, 0)

# =================================
# Circuit B: 1 occurs more frequently
# =================================
circuit_B = QuantumCircuit(1, 1)
circuit_B.ry(2.5, 0)
circuit_B.measure(0, 0)

# =================================
# Execute circuits for 100 shots
# =================================
circuits = [circuit_A, circuit_B]
names = ["Circuit A: 0 More Frequent",
         "Circuit B: 1 More Frequent"]

results = {}

for name, circuit in zip(names, circuits):

    print("\n", name)
    print("Quantum Circuit:")
    print(circuit.draw())

    job = simulator.run(circuit, shots=100)
    counts = job.result().get_counts()

    results[name] = counts

    print("Predicted probabilities:")
    if "Circuit A" in name:
        print("0: Approximately 94%, 1: Approximately 6%")
    else:
        print("0: Approximately 10%, 1: Approximately 90%")

    print("Actual measurement counts:", counts)

    plot_histogram(counts, title=name)
    plt.show()

# =================================
# Compare both circuits
# =================================
print("\nRESULT COMPARISON")
print("Circuit A: Outcome 0 should occur more frequently.")
print("Circuit B: Outcome 1 should occur more frequently.")

# =================================
# Challenge Question Answer
# =================================
print("\nCHALLENGE QUESTION ANSWER:")
print("Quantum gates change the quantum state of a qubit.")
print("This changes the probabilities of measuring 0 or 1.")
print("The Ry gate rotates the qubit by a chosen angle.")
print("Different rotation angles produce different probabilities.")
```
